Pandas will read the dataset in format of CSV.
x : features (inputs)
":" take all the columns
": - 1" take all the columns except the last one.
so x contains all the input variables (features) used to make predictions. in this case is region, age, and income

y: target (output/label)
"-1" take the last column only
so y contains the output/label you want to predict. in this case is online shopper.
